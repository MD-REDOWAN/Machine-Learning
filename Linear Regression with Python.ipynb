```python
import pandas as pd
from matplotlib import pyplot as plt
```

```python
df = pd.read_csv('nasdaq100.csv',sep=';')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df = df.drop(columns= 'Date')
```

```python
df.head()
```

```python
plt.scatter(df['Starting (USD)'], df['Ending (USD)'])
plt.xlabel('Starting (USD)')
plt.ylabel('Ending (USD)')
plt.title("Stock Prices of Dataset made by 'Redowan;")
```

```python
x = df.drop('Ending (USD)', axis=1)
```

```python
x.head()
```

```python
y = df['Ending (USD)']
y.head()
```

Linear Regression


```python
!pip install scikit-learn
```

```python
from sklearn.linear_model import LinearRegression
```

```python
reg = LinearRegression()
```

```python
x.mean()
```

```python
y.mean()
```

```python
plt.scatter(x.mean() , y.mean(),color = 'red')
plt.scatter(df['Starting (USD)'], df['Ending (USD)'])
plt.xlabel('Starting (USD)')
plt.ylabel('Ending (USD)')
plt.title("Stock Prices of Dataset made by 'Redowan;")
```

Model Fit (Train)

```python
# Train the ml model
reg.fit(x,y)
```

```python
m = reg.coef_
```

```python
c = reg.intercept_
```

```python
# y = mx+c   
# # x input()
# output = predicted Result

m*16700 + c
```

```python
reg.predict[16700]
```

```python
df['predicted_y'] = reg.predict(x)
df.head()
```

```python
plt.plot(x, df['predicted_y'],color = 'red')
```

```python
# Another way to predicted y
plt.plot(x, reg.predict(x))
```
