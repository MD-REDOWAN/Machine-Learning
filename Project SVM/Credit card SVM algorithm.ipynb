```python
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(x,y,test_size=.25,random_state=1)
```

```python
model = SVC(gamma='auto')
```

```python
import pandas as pd
import numpy as np

print((xtrain == 'missing').sum())
```

```python
# Replace the string 'missing' with actual NaN values
xtrain = xtrain.replace('missing', np.nan)
xtest = xtest.replace('missing', np.nan)

# Fill missing numeric values with each column's median
for col in xtrain.select_dtypes(include=np.number).columns:
    median_value = xtrain[col].median()
    xtrain[col] = xtrain[col].fillna(median_value)
    xtest[col] = xtest[col].fillna(median_value)
```

```python
print(xtrain.select_dtypes(include=['object', 'category']).columns.tolist())
```

```python
xtrain = pd.get_dummies(xtrain, drop_first=True, dtype=int)
xtest = pd.get_dummies(xtest, drop_first=True, dtype=int)

# Ensure train and test have matching columns
xtest = xtest.reindex(columns=xtrain.columns, fill_value=0)
```

```python
model.fit(xtrain,ytrain)
```

```python
model.score(xtest,ytest)
```
