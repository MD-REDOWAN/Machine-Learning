Statistical Analysis

```python
# Statistic analysis for numarical columns
df.describe()
```

```python
# Correlation

import seaborn as sns
# Pass numeric_only=True to df.corr()
sns.heatmap(df.corr(numeric_only=True), annot=True) 
# numeric_only=True tells Pandas to automatically ignore text columns (like your country column containing 'US', names, or categories) and only calculate correlation for integers and floats.
```

```python
# Subseting rows for neumerical values
df['points']>30
```

```python
# subseting show the full dataset df[df]
df[df['points'] >30 ]
```

```python
# Subseting for sting values

df['country'] == 'Italy'
```

```python
# show the hole dataset
df[df['country'] == 'Italy']
```

```python
# OR funtion(|)
df[(df['country']=='Italy') | (df['country'] == 'spain')]
```

```python
# using ( .isin) function
df[df["country"] .isin(['Italy', 'spain'])]
```
